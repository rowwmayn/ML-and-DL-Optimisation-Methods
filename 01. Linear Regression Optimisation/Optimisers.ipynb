```python
df = pd.read_csv(r"F:\Personal Projects\ML-and-DL-Optimisation-Methods\Datasets\housing.csv", sep=r"\s+", engine="python", encoding="utf-8");
#rint(df.head(10));
data_matrix = df.to_numpy();
print(data_matrix.shape);
r,c = data_matrix.shape;
X = data_matrix[:,0:-1];
#print(X.shape);
Y = data_matrix[:,-1];
Design_X = np.hstack((np.ones((r,1)),X));
print(Design_X);


X_train, X_test, Y_train, Y_test = tts(X,Y,train_size=0.8);
print(Y_train.shape)
```

```python
def VGD(X,Y,step_size=0.0000001, iterations=500000):
    
    thetas = [];
    costs = [];

    m,n = X.shape;
    theta = np.zeros((n,1));
    Y = Y.reshape(m,1);
    
    

    for i in range(iterations):

        y_pred = X@theta;
        error = y_pred - Y;
        sqrd_error = np.square(error);
        cost = (1/(2*m))*np.sum(sqrd_error)
        grad = (1/m)*(X.T@(error));
        theta = theta - step_size*grad;
        thetas.append(theta.copy());
        costs.append(cost.copy());

    return theta, thetas, costs;


```

```python
theta, thetas, costs = VGD(X_train,Y_train);
```

```python
def MAE(theta, X_test, Y_test):
    mae = np.mean((np.abs((X_test@theta)-Y_test)));
    return mae;
```
